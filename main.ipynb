```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
data = pd.read_csv("data.csv")
```

```python
data.info()
```

```python
data.head()
```

```python
data.describe()
```

```python
#Dropping Column of NAN and ID
data.drop(['Unnamed: 32', "id"], axis=1, inplace=True)

#Verifying that data was dropped
sns.heatmap(data.isnull(),cmap = "viridis")
```

```python
data.diagnosis = [1 if value == "M" else 0 for value in data.diagnosis]
data.diagnosis = data.diagnosis.astype("category",copy = False)
```

```python
#Splitting data

x = data.drop("diagnosis", inplace= False, axis = 1)
y = data["diagnosis"]
```

```python
#normalizing data
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

x_scaled
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled,y,test_size=0.30, random_state= 42)
```

```python
#Training and Testing Model
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt


acc = accuracy_score(y_test,y_pred)
print(f"The Accuracy of this model is: {acc:.2}")

pre = precision_score(y_test,y_pred)
print(f"The Precision of this model is:  {pre:.2}")

recall = recall_score(y_test,y_pred)
print(f"The Recall Score of this model is: {recall:.2}")

f1 = f1_score(y_test,y_pred)
print(f"The F1-Score of this model is: {f1:.2}")

cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap = "Blues")
plt.title("Confusion Matrix")
plt.show()
```

```python
#Testing for Over fitting

y_train_pred = model.predict(x_train)
cm_train = confusion_matrix(y_train, y_train_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm_train)
disp.plot(cmap = "Blues")
plt.title("Training Confusion Matrix")
plt.show()
```

## The Model is neither Overfit nor Underfit as shown by the aligned accuracies on both training and testing datasets, This shows good generalization
